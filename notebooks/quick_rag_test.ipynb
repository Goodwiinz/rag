```python
# Install required packages
!pip install requests pandas matplotlib plotly --quiet

import requests
import json
import time
import pandas as pd
from datetime import datetime
from IPython.display import display, HTML

# Configuration
BASE_URL = "http://localhost:8000"
session = requests.Session()
auth_token = None

print("🚀 Quick RAG System Test Initialized!")
print(f"📍 Backend: {BASE_URL}")
print(f"🌐 Frontend: http://localhost:3000")
```

## 🏥 1. System Health Check

```python
print("🏥 Checking system health...")

try:
    response = session.get(f"{BASE_URL}/health", timeout=5)
    if response.status_code == 200:
        data = response.json()
        print(f"✅ API Status: {data.get('status')}")
        print(f"📊 Version: {data.get('version')}")
        print(f"🌍 Environment: {data.get('environment')}")
        system_healthy = True
    else:
        print(f"❌ API unhealthy: {response.status_code}")
        system_healthy = False
except Exception as e:
    print(f"❌ Health check failed: {str(e)}")
    system_healthy = False

if not system_healthy:
    print("\n🔧 **Troubleshooting:**")
    print("1. Run: `docker-compose up -d`")
    print("2. Check: `docker-compose ps`")
    print("3. Logs: `docker-compose logs -f backend`")
```

## 🔐 2. User Authentication

```python
if system_healthy:
    test_timestamp = int(time.time())
test_email = f"test_{test_timestamp}@example.com"
test_password = "SecurePass123!"
test_org = f"Test Org {test_timestamp}"

print(f"🔐 Testing authentication...")
print(f"   Email: {test_email}")
print(f"   Org: {test_org}\n")

reg_data = {
    "email": test_email,
    "password": test_password,
    "first_name": "Test",
    "last_name": "User",
    "organization_name": test_org
}

reg_response = session.post(f"{BASE_URL}/api/v1/auth/register", json=reg_data, timeout=10)
if reg_response.status_code in [200, 201]:
    print("✅ Registration successful!")
    
    login_data = {"email": test_email, "password": test_password}
    login_response = session.post(f"{BASE_URL}/api/v1/auth/login", json=login_data, timeout=10)
    
    if login_response.status_code == 200:
        login_result = login_response.json()
        auth_token = login_result.get("access_token")
        session.headers.update({"Authorization": f"Bearer {auth_token}"})
        print("✅ Login successful!")
        print(f"🔑 Token received\n")
        auth_success = True
    else:
        print(f"❌ Login failed: {login_response.status_code}")
        auth_success = False
else:
    print(f"❌ Registration failed: {reg_response.status_code}")
    print(f"Response: {reg_response.text[:200]}")
    auth_success = False
```

## 📄 3. Document Upload Test

```python
# ========================================
# COMPLETE AUTH + UPLOAD TEST (FIXED!)
# ========================================

import time

print("=" * 60)
print("🔐 STEP 1: Authentication")
print("=" * 60)

# Generate unique credentials
test_timestamp = int(time.time())
test_email = f"test_{test_timestamp}@example.com"
test_password = "SecurePass123!"
test_org = f"Test Org {test_timestamp}"

print(f"📧 Email: {test_email}")
print(f"🏢 Org: {test_org}\n")

# Register
reg_data = {
    "email": test_email,
    "password": test_password,
    "first_name": "Test",
    "last_name": "User",
    "organization_name": test_org
}

reg_response = session.post(f"{BASE_URL}/api/v1/auth/register", json=reg_data, timeout=10)
if reg_response.status_code in [200, 201]:
    print("✅ Registration successful")
else:
    print(f"❌ Registration failed: {reg_response.status_code}")
    print(reg_response.text[:300])

# Login
login_data = {"email": test_email, "password": test_password}
login_response = session.post(f"{BASE_URL}/api/v1/auth/login", json=login_data, timeout=10)

if login_response.status_code == 200:
    login_result = login_response.json()
    auth_token = login_result.get("access_token")
    
    # Set authorization header
    session.headers.update({"Authorization": f"Bearer {auth_token}"})
    
    print("✅ Login successful")
    print(f"🔑 Token received and set\n")
    
    print("=" * 60)
    print("📄 STEP 2: Document Upload")
    print("=" * 60)
    
    # Create test document
    sample_content = """# Artificial Intelligence Overview

## Introduction
Artificial Intelligence (AI) is transforming how we interact with technology.

### Machine Learning
- Supervised learning algorithms
- Deep neural networks

### Applications
- Natural language processing
- Computer vision
"""
    
    with open("test_ai_doc.txt", "w") as f:
        f.write(sample_content)
    
    print("📝 Created test document\n")
    
    # Upload document
    with open("test_ai_doc.txt", "rb") as f:
        files = {"file": ("test_ai_doc.txt", f, "text/plain")}
        data = {"title": "AI Overview Test", "description": "Test document"}
        
        upload_response = session.post(
            f"{BASE_URL}/api/v1/files/upload",
            files=files,
            data=data,
            timeout=30
        )
    
    if upload_response.status_code in [200, 201]:
        upload_result = upload_response.json()
        print(f"✅ Upload successful!")
        print(f"📄 Document ID: {upload_result.get('id')}")
        print(f"📊 Status: {upload_result.get('processing_status')}")
        auth_success = True
        upload_success = True
    else:
        print(f"❌ Upload failed: {upload_response.status_code}")
        print(upload_response.text[:300])
        upload_success = False
else:
    print(f"❌ Login failed: {login_response.status_code}")
    auth_success = False

print("\n" + "=" * 60)
print("✅ TEST COMPLETE!")
print("=" * 60)
```

## 🔍 4. Search Functionality Test

```python
if auth_success:
    print("🔍 Testing search functionality...")
    
    search_queries = [
        "Artificial Intelligence",
        "Machine Learning",
        "Deep Learning",
        "Neural Networks"
    ]
    
    search_results = []
    
    for query in search_queries:
        print(f"\n🔎 Searching: '{query}'")
        
        search_data = {"query": query, "limit": 5}
        
        try:
            # Updated API path to /api/v1/*
            search_response = session.post(f"{BASE_URL}/api/v1/search/hybrid", json=search_data, timeout=15)
            
            if search_response.status_code == 200:
                result = search_response.json()
                results = result.get("results", [])
                
                search_results.append({
                    "query": query,
                    "results_count": len(results),
                    "total": result.get("total", 0),
                    "response_time": result.get("search_time", 0)
                })
                
                print(f"✅ Found {len(results)} results in {result.get('search_time', 0):.3f}s")
                
                if results:
                    for i, res in enumerate(results[:2], 1):
                        print(f"   {i}. {res.get('title', 'Unknown')} (Score: {res.get('score', 0):.3f})")
                
            else:
                print(f"❌ Search failed: {search_response.status_code}")
                print(f"   Response: {search_response.text[:200]}")
                search_results.append({"query": query, "results_count": 0, "error": str(search_response.status_code)})
                
        except Exception as e:
            print(f"❌ Search error: {str(e)}")
            search_results.append({"query": query, "results_count": 0, "error": str(e)})
    
    # Create search results table
    if search_results:
        df_search = pd.DataFrame([
            {
                "Query": r["query"],
                "Results": r.get("results_count", 0),
                "Total": r.get("total", 0),
                "Time (s)": f"{r.get('response_time', 0):.3f}",
                "Status": "✅ Success" if r.get("results_count", 0) > 0 else "❌ No Results"
            }
            for r in search_results
        ])
        
        print("\n📊 **Search Results Summary:**")
        display(df_search)
        
        # Calculate search statistics
        successful_searches = [r for r in search_results if r.get("results_count", 0) > 0]
        print(f"\n📈 **Search Statistics:**")
        print(f"- Successful Searches: {len(successful_searches)}/{len(search_queries)}")
        print(f"- Success Rate: {(len(successful_searches)/len(search_queries))*100:.1f}%")
        
        if successful_searches:
            avg_results = sum(r.get("results_count", 0) for r in successful_searches) / len(successful_searches)
            avg_time = sum(r.get("response_time", 0) for r in successful_searches) / len(successful_searches)
            print(f"- Average Results per Query: {avg_results:.1f}")
            print(f"- Average Response Time: {avg_time:.3f}s")
    
else:
    print("❌ Skipping search - authentication required")
```

## 🔒 6. Security Features Test

```python
if auth_success:
    print("🔒 Testing security features...")
    print("ℹ️  Note: Security endpoints are currently being refactored\n")
    
    # Test encryption status
    try:
        enc_response = session.get(f"{BASE_URL}/api/v1/encryption/status", timeout=10)
        
        if enc_response.status_code == 200:
            enc_data = enc_response.json()
            print("✅ Encryption service accessible")
            
            key_mgmt = enc_data.get("key_management", {})
            if key_mgmt:
                print(f"🔑 Key Management:")
                data_key = key_mgmt.get("data", {})
                print(f"   - Data Key: {data_key.get('active_key_id', 'N/A')}")
                print(f"   - Algorithm: {data_key.get('key_algorithm', 'N/A')}")
        elif enc_response.status_code == 404:
            print(f"⏳ Encryption service: Not yet implemented (expected)")
        else:
            print(f"⚠️ Encryption service: {enc_response.status_code}")
    except Exception as e:
        print(f"⏳ Encryption service: Not yet available")
    
    # Test audit logs
    try:
        audit_response = session.get(f"{BASE_URL}/api/v1/compliance/audit/logs", params={"limit": 5}, timeout=10)
        
        if audit_response.status_code == 200:
            audit_data = audit_response.json()
            print(f"✅ Audit logs retrieved: {len(audit_data)} entries")
            
            # Show recent operations
            operations = {}
            for log in audit_data:
                op_type = log.get("operation_type", "Unknown")
                operations[op_type] = operations.get(op_type, 0) + 1
            
            if operations:
                print(f"📋 Recent Operations:")
                for op, count in operations.items():
                    print(f"   - {op}: {count}")
        elif audit_response.status_code == 404:
            print(f"⏳ Audit logs: Not yet implemented (expected)")
        else:
            print(f"⚠️ Audit logs: {audit_response.status_code}")
    except Exception as e:
        print(f"⏳ Audit logs: Not yet available")
    
    # Test RBAC
    try:
        rbac_response = session.get(f"{BASE_URL}/api/v1/rbac/permissions", timeout=10)
        
        if rbac_response.status_code == 200:
            rbac_data = rbac_response.json()
            print("✅ RBAC permissions accessible")
            print(f"🛡️ Permissions available: {len(rbac_data) if isinstance(rbac_data, list) else 'Unknown'}")
        elif rbac_response.status_code == 404:
            print(f"⏳ RBAC permissions: Not yet implemented (expected)")
        else:
            print(f"⚠️ RBAC permissions: {rbac_response.status_code}")
    except Exception as e:
        print(f"⏳ RBAC permissions: Not yet available")
    
    print("\nℹ️  Security features are being refactored and will be available in a future update.")
    
else:
    print("❌ Skipping security tests - authentication required")
```
